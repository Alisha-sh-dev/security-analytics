# Scenario 1 – Loan Eligibility Evaluation System

```python
# ============================================
# SCENARIO 1: Loan Eligibility Evaluation System
# ============================================

print("=" * 50)
print("     LOAN ELIGIBILITY EVALUATION SYSTEM")
print("=" * 50)

# Get user inputs
print("\nPlease enter the following information:")
print("-" * 40)

salary = float(input("Monthly Salary (AED): "))
employment_years = float(input("Years of Employment: "))
credit_score = int(input("Credit Score (300-850): "))
existing_loan = input("Existing Loan? (Yes/No): ").strip().lower()

print("\n" + "=" * 50)
print("EVALUATING YOUR APPLICATION...")
print("=" * 50)

# Nested if statements for loan decision
if existing_loan == "yes":
    # Check if they have an existing loan but still might qualify
    if salary >= 15000 and employment_years >= 2 and credit_score >= 700:
        decision = "APPROVED"
        explanation = "You meet all criteria despite having an existing loan."
    elif 10000 <= salary <= 14999 and credit_score >= 750:
        decision = "CONDITIONAL APPROVAL"
        explanation = "Your strong credit score qualifies you, but salary is moderate."
    else:
        decision = "REJECTED"
        explanation = "Existing loan combined with insufficient financial metrics."
else:
    # No existing loan - standard evaluation
    if salary >= 15000 and employment_years >= 2 and credit_score >= 700:
        decision = "APPROVED"
        explanation = f"Excellent! Salary: {salary:.2f} AED, Employment: {employment_years} years, Credit Score: {credit_score}"
    elif 10000 <= salary <= 14999 and credit_score >= 750:
        decision = "CONDITIONAL APPROVAL"
        explanation = f"Your credit score ({credit_score}) is excellent, but salary ({salary:.2f} AED) needs improvement."
    else:
        decision = "REJECTED"
        # Determine specific rejection reason
        if salary < 10000:
            explanation = f"Salary ({salary:.2f} AED) is below minimum requirement of 10,000 AED."
        elif 10000 <= salary <= 14999 and credit_score < 750:
            explanation = f"Credit score ({credit_score}) is below 750 required for conditional approval."
        elif salary >= 15000 and (employment_years < 2 or credit_score < 700):
            if employment_years < 2:
                explanation = f"Employment years ({employment_years}) is less than 2 years required."
            else:
                explanation = f"Credit score ({credit_score}) is below 700 required."
        else:
            explanation = "You do not meet the minimum requirements for approval."

# Display results
print(f"\n{' FINAL DECISION ':-^50}")
print(f"Loan Decision: {decision}")
print(f"Explanation: {explanation}")
print("=" * 50)
```

# Scenario 2 – Retail Billing System

```python
# ============================================
# SCENARIO 2: Retail Billing System
# ============================================

print("=" * 50)
print("        RETAIL BILLING SYSTEM")
print("=" * 50)

# Get customer information
print("\n--- Customer Information ---")
customer_type = input("Customer Type (Regular / Member / VIP): ").strip().lower()
num_items = int(input("Number of items purchased: "))

# Initialize variables
total_amount = 0
item_prices = []

print("\n--- Enter Item Prices ---")
# Loop to enter price of each item
for i in range(1, num_items + 1):
    price = float(input(f"Enter price for item {i}: AED "))
    item_prices.append(price)
    total_amount += price
    print(f"  ✓ Item {i}: AED {price:.2f} (Running total: AED {total_amount:.2f})")

print("\n" + "=" * 50)
print("CALCULATING BILL...")
print("=" * 50)

# Display itemized bill
print("\n--- Itemized Bill ---")
for i, price in enumerate(item_prices, 1):
    print(f"Item {i}: AED {price:.2f}")

print(f"\nTotal Purchase Amount: AED {total_amount:.2f}")

# Calculate discount based on rules
discount_percentage = 0
additional_discount = 0

# Base discount based on customer type
if customer_type == "regular":
    discount_percentage = 3
elif customer_type == "member":
    discount_percentage = 5
elif customer_type == "vip":
    discount_percentage = 7
else:
    print(f"\n⚠ Warning: '{customer_type}' is not a valid customer type. No discount applied.")
    discount_percentage = 0

# Check if total is 500 or more to apply base discount
if total_amount >= 500:
    base_discount = total_amount * (discount_percentage / 100)
    print(f"\nBase Discount ({discount_percentage}% for {customer_type.title()} customer): AED {base_discount:.2f}")
else:
    base_discount = 0
    print(f"\nNo discount applied (minimum AED 500 required).")

# Additional 5% discount if total is 1000 or more
if total_amount >= 1000:
    additional_discount = total_amount * 0.05
    print(f"Additional Discount (5% for purchases ≥ AED 1000): AED {additional_discount:.2f}")

# Calculate total discount and final amount
total_discount = base_discount + additional_discount
final_amount = total_amount - total_discount

print("\n" + "-" * 40)
print(f"Total Discount: AED {total_discount:.2f}")
print(f"FINAL PAYABLE AMOUNT: AED {final_amount:.2f}")
print("=" * 50)

# Thank you message
print(f"\nThank you for shopping with us, {customer_type.title()} customer!")
print("Have a great day!")
```

# Scenario 3 – Employee Performance Evaluation System

```python
# ============================================
# SCENARIO 3: Employee Performance Evaluation System
# ============================================

# Function to evaluate employee performance
def evaluate_performance(sales_amount, satisfaction_score, attendance_percentage):
    """
    Evaluates employee performance based on three criteria.

    Parameters:
    - sales_amount: float - Total sales in AED
    - satisfaction_score: float - Customer satisfaction (0-100)
    - attendance_percentage: float - Attendance percentage (0-100)

    Returns:
    - grade: str - 'A', 'B', or 'C'
    - bonus_percentage: float - Bonus percentage (20%, 10%, or 0%)
    - criteria_met: int - Number of criteria satisfied
    """

    # Check each criterion
    sales_met = sales_amount >= 100000
    satisfaction_met = satisfaction_score >= 85
    attendance_met = attendance_percentage >= 90

    # Count how many criteria are satisfied
    criteria_met = sum([sales_met, satisfaction_met, attendance_met])

    # Determine grade and bonus based on criteria met
    if criteria_met == 3:
        grade = "A"
        bonus_percentage = 20
    elif criteria_met == 2:
        grade = "B"
        bonus_percentage = 10
    else:
        grade = "C"
        bonus_percentage = 0

    return grade, bonus_percentage, criteria_met


# Main program
print("=" * 60)
print("     EMPLOYEE PERFORMANCE EVALUATION SYSTEM")
print("=" * 60)

# Initialize counters for summary
grade_a_count = 0
grade_b_count = 0
grade_c_count = 0

# List to store all employee results
employee_results = []

# Get number of employees to process
num_employees = int(input("\nHow many employees would you like to evaluate? "))

print("\n" + "=" * 60)
print("BEGINNING EVALUATION...")
print("=" * 60)

# Loop to process multiple employees
for emp_num in range(1, num_employees + 1):
    print(f"\n--- Employee #{emp_num} ---")

    # Get employee information
    name = input("Employee Name: ")

    # Input validation for sales
    while True:
        try:
            sales = float(input("Sales Amount (AED): "))
            if sales < 0:
                print("Sales cannot be negative. Please re-enter.")
                continue
            break
        except ValueError:
            print("Invalid input. Please enter a number.")

    # Input validation for satisfaction score
    while True:
        try:
            satisfaction = float(input("Customer Satisfaction Score (0-100): "))
            if 0 <= satisfaction <= 100:
                break
            else:
                print("Score must be between 0 and 100.")
        except ValueError:
            print("Invalid input. Please enter a number.")

    # Input validation for attendance
    while True:
        try:
            attendance = float(input("Attendance Percentage (0-100): "))
            if 0 <= attendance <= 100:
                break
            else:
                print("Percentage must be between 0 and 100.")
        except ValueError:
            print("Invalid input. Please enter a number.")

    # Call the function to evaluate performance
    grade, bonus, criteria_met = evaluate_performance(sales, satisfaction, attendance)

    # Store results
    employee_results.append({
        'name': name,
        'grade': grade,
        'bonus': bonus,
        'criteria_met': criteria_met,
        'sales': sales,
        'satisfaction': satisfaction,
        'attendance': attendance
    })

    # Update counters
    if grade == 'A':
        grade_a_count += 1
    elif grade == 'B':
        grade_b_count += 1
    else:
        grade_c_count += 1

    # Display individual employee result
    print("\n" + "-" * 40)
    print(f"📊 RESULTS FOR {name.upper()}:")
    print(f"   Performance Grade: {grade}")
    print(f"   Bonus Percentage: {bonus}%")
    print(f"   Criteria Met: {criteria_met}/3")
    if bonus > 0:
        bonus_amount = sales * (bonus / 100)
        print(f"   Bonus Amount: AED {bonus_amount:,.2f}")
    print("-" * 40)

# Display final summary
print("\n" + "=" * 60)
print("              FINAL SUMMARY REPORT")
print("=" * 60)

print("\n--- Individual Results ---")
for emp in employee_results:
    print(f"• {emp['name']}: Grade {emp['grade']} ({emp['bonus']}% bonus)")

print("\n--- Grade Distribution ---")
print(f"🏆 Grade A Employees (20% bonus): {grade_a_count}")
print(f"⭐ Grade B Employees (10% bonus): {grade_b_count}")
print(f"📋 Grade C Employees (No bonus): {grade_c_count}")
print(f"\nTotal Employees Evaluated: {num_employees}")

# Calculate percentages
if num_employees > 0:
    print(f"\n--- Performance Breakdown ---")
    print(f"Grade A: {(grade_a_count/num_employees)*100:.1f}%")
    print(f"Grade B: {(grade_b_count/num_employees)*100:.1f}%")
    print(f"Grade C: {(grade_c_count/num_employees)*100:.1f}%")

print("\n" + "=" * 60)
print("Evaluation Complete!")
print("=" * 60)
```
