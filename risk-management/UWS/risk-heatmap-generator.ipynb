```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Data
data = {
    'Risk ID': ['R1', 'R2', 'R3', 'R4', 'R5', 'R6'],
    'Risk Name': [
        'SQL Injection Attack',
        'Ransomware Attack',
        'Data Breach or Leakage',
        'Data Loss or Corruption',
        'Denial of Service Attack',
        'Phishing or Social Engineering Attack',
    ],
    'Risk Likelihood': [
        'Low',
        'Low to Medium (depending on the security posture)',
        'Low to Medium (depending on the security posture)',
        'Low to Medium (depending on the backup and recovery systems)',
        'Low to Medium (depending on the network security)',
        'Medium to High (depending on the user awareness and training)',
    ],
    'Risk Impact': [
        'High',
        'High to Critical (depending on the data)',
        'High to Critical (depending on the data)',
        'Medium to High (depending on the data)',
        'Medium to High (depending on the service)',
        '-',
    ],
    'Risk Score': [4, 'Medium to High (depending on the data)', 'Medium to High (depending on the data)', 'Medium', 'Medium', 'Medium'],
    'Risk Level': ['Medium', 'Medium to High (depending on the data)', 'Medium to High (depending on the data)', 'Medium', 'Medium', '-'],
    'Risk Color': ['Yellow', 'Yellow to Red (depending on the data)', 'Yellow to Red (depending on the data)', 'Yellow', 'Yellow', 'Yellow'],
}
```

```python
# Create a DataFrame
df = pd.DataFrame(data)

# Create a function to map risk levels to corresponding colors
def map_color(level):
    if 'Red' in level:
        return 3
    elif 'Yellow' in level:
        return 2
    elif 'Medium' in level:
        return 1
    else:
        return 0  # You can define colors for other levels as needed

# Apply the mapping function to 'Risk Color' and create a new column 'Color'
df['Color'] = df['Risk Color'].apply(map_color)

# Pivot the data to prepare it for the heatmap
heatmap_data = df.pivot('Risk Name', 'Risk ID', 'Color')

# Create the heatmap
plt.figure(figsize=(10, 6))
cmap = sns.color_palette(['green', 'yellow', 'red'])
ax = sns.heatmap(heatmap_data, cmap=cmap, annot=True, fmt="", cbar=False)
ax.set_xticklabels(ax.get_xticklabels(), rotation=45)
plt.title('Risk Heatmap')
plt.show()
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Data
data = {
    'Threat ID': ['T1', 'T2', 'T3', 'T4', 'T5', 'T6'],
    'Threat Name': [
        'SQL Injection Attack',
        'Ransomware Attack',
        'Data Breach or Leakage',
        'Data Loss or Corruption',
        'Denial of Service Attack',
        'Phishing or Social Engineering Attack',
    ],
    'Damage Potential': [8, 9, 10, 8, 7, 6],
    'Reproducibility': [8, 7, 6, 5, 7, 8],
    'Exploitability': [7, 6, 5, 4, 6, 8],
    'Affected Users': [9, 10, 10, 9, 8, 7],
    'Discoverability': [6, 5, 4, 3, 7, 8],
    'DREAD Score': [7.6, 7.4, 7, 5.8, 7, 7.4],
}

# Create a DataFrame
df = pd.DataFrame(data)

# Sort the DataFrame by DREAD Score
df = df.sort_values(by='DREAD Score', ascending=False)

# Create a bar plot for DREAD Scores
plt.figure(figsize=(10, 6))
sns.barplot(data=df, x='DREAD Score', y='Threat Name', palette="viridis")
plt.xlabel('DREAD Score (0-10)')
plt.title('Threat Analysis Model')
plt.show()
```
